```python
df["Reorder_Risk"] = np.where(
    df["Inventory_Level"] <= df["Reorder_Point"],
    "Reorder Required",
    "Sufficient Stock"
)
```

```python
df["Reorder_Risk"].value_counts()
```

```python
df["Reorder_Risk"].value_counts(normalize=True) * 100
```

```python
df["Inventory_Value"] = (
    df["Inventory_Level"] * df["Unit_Cost"]
)
```

```python
df["Inventory_Sales_Value"] = (
    df["Inventory_Level"] * df["Unit_Price"]
)
```

```python
df["Gross_Margin_Per_Unit"] = (
    df["Unit_Price"] - df["Unit_Cost"]
)
```

```python
df["Gross_Margin_Percentage"] = (
    df["Gross_Margin_Per_Unit"] / df["Unit_Price"] * 100
)
```

```python
df["Days_of_Inventory"] = np.where(
    df["Demand_Forecast"] > 0,
    df["Inventory_Level"] / df["Demand_Forecast"],
    np.nan
)
```

```python
df["Inventory_Status"] = np.select(
    [
        df["Days_of_Inventory"] < 7,
        df["Days_of_Inventory"].between(7, 30),
        df["Days_of_Inventory"] > 30
    ],
    [
        "High Risk",
        "Healthy",
        "Overstock"
    ],
    default="Unknown"
)
```

```python
import os

os.makedirs("../data/cleaned", exist_ok=True)

output_path = "../data/cleaned/supply_chain_cleaned.csv"

df.to_csv(output_path, index=False)

print("Saved:", output_path)
```

```python
df["Inventory_Status"] = np.select(
    [
        df["Demand_Forecast"] == 0,
        df["Days_of_Inventory"] < 7,
        df["Days_of_Inventory"].between(7, 30),
        df["Days_of_Inventory"] > 30
    ],
    [
        "No Forecast Demand",
        "High Risk",
        "Healthy",
        "Overstock"
    ],
    default="Unknown"
)
```
